```python
from pandas.plotting import scatter_matrix
from sklearn.datasets import load_wine
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

wine = load_wine()

X = wine.data
y = wine.target

df_wine = pd.DataFrame(X, columns=wine.feature_names)
df_wine['target'] = y

df_wine.head()
```

```python
print("Quantidade de amostras:", df_wine.shape[0])
print("Quantidade de features:", df_wine.shape[1] - 1)
print("Quantidade de classes:", df_wine['target'].nunique())
print("Classes:", df_wine['target'].unique())
```

```python
sns.pairplot(df_wine, hue='target')

plt.show()
```

```python
X = df_wine.drop('target', axis=1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)

df_pca_2d = pd.DataFrame(X_pca_2d, columns=['PC1', 'PC2'])
df_pca_2d['target'] = y

df_pca_2d.head()
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df_pca_2d,
    x='PC1',
    y='PC2',
    hue='target',
    palette='Set1'
)

plt.title('PCA - 2 Componentes Principais')
plt.show()
```

```python
print("Variância explicada por componente:")
print(pca_2d.explained_variance_ratio_)
print("\nVariância explicada acumulada:")
print(pca_2d.explained_variance_ratio_.sum())
```

```python
componentes = pd.DataFrame(
    pca_2d.components_,
    columns=X.columns,
    index=['PC1', 'PC2']
)

componentes
```

```python
print("Maiores coeficientes absolutos da PC1:")
print(componentes.loc['PC1'].abs().sort_values(ascending=False).head())

print("\n")

print("Maiores coeficientes absolutos da PC2:")
print(componentes.loc['PC2'].abs().sort_values(ascending=False).head())
```

```python
X = df_wine.drop('target', axis=1)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA()

pca.fit(X_scaled)

variancia_acumulada = np.cumsum(
    pca.explained_variance_ratio_
)

plt.figure(figsize=(8,6))

plt.plot(
    range(1, len(variancia_acumulada)+1),
    variancia_acumulada,
    marker='o'
)

plt.xlabel('Número de Componentes Principais')
plt.ylabel('Variância Explicada Acumulada')
plt.title('Elbow Plot - PCA Wine Dataset')

plt.grid(True)

plt.show()
```
